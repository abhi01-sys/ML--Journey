```python
from sklearn.datasets import load_diabetes

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
reg.fit(X_train,y_train)
```

```python
y_pred = reg.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```

```python
class MeraMLR:

  def __init__(self):
    self.coef_ = None
    self.intercept = None

  def fit(self,X_train,Y_train):
    X_train = np.insert(X_train,0,1,axis=1)

    # calculate coefficient

    betas = np.linalg.inv(np.dot(X_train.T,X_train)).dot(X_train.T).dot(Y_train)
    self.intercept = betas[0]
    self.coef_ = betas[1:]
    print(self.intercept)
    print(self.coef_)
    print(betas)

  def predict(self,X_test):
    y_pred = np.dot(X_test,self.coef_) + self.intercept
    return y_pred


```

```python
#np.insert(a,b,c,d) is used to insert an new column to a dataframe
# a -> in which data you have to insert , b-> which column , c -> insert value , d -> in row or column

#numpy.linalg.inv() -> to calculate inverse

#X_train.T -> gives transpose of X-train
```

```python
lr = MeraMLR()
```

```python
lr.fit(X_train,y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```
