# One Hot Encoding/Nominal

```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder
```

```python
df = sns.load_dataset('tips')
```

```python
df
```

```python
encoder = OneHotEncoder()
```

```python
encoded = encoder.fit_transform(df[['sex']])
```

```python
encoder_df = pd.DataFrame(encoded.toarray(),columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
encoder.transform([['Male']]).toarray()
```

```python
pd.concat([df,encoder_df],axis=1)
```

#Label Encoding and Ordered Encoding

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()
```

```python
encoded = encoder.fit_transform(df['day'])
```

```python
encoded_df = pd.DataFrame(encoded,columns=['day'])
```

```python
encoded_df
```

```python
encoder.transform(['Sun', 'Sat', 'Thur', 'Fri'])
```

```python
pd.concat([df,encoded_df],axis=1)
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
Ordinal_EN = OrdinalEncoder(categories=[['Thur','Fri','Sat','Sun']])
```

```python
encoder = Ordinal_EN.fit_transform(df[['day']])
```

```python
encoded_df = pd.DataFrame(encoder,columns=['day'])
```

```python
encoded_df
```

```python
final_df = pd.concat([df,encoded_df],axis=1)
```

```python
final_df
```

#Target Guided Ordinal Encoding

```python
new_df = sns.load_dataset('tips')
```

```python
from sklearn.preprocessing import TargetEncoder
```

```python
encoder = TargetEncoder()
```

```python
encoded = encoder.fit_transform(new_df[['day']],new_df['total_bill'])
```

```python
df_encoded = pd.DataFrame(encoded,columns=['day_encoded_total_bill'])
```

```python
pd.concat([new_df,df_encoded],axis=1)
```
