```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_regression
```

```python
X,y = make_regression(n_samples=1000,n_features=13,n_informative=7,random_state=13,noise=20)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.linear_model import ElasticNet
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2 ,random_state=13)
```

```python
lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
r2_score(y_test,y_pred)
```

```python
reg = Ridge(alpha = 0.0085)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
def rdge(alpha):

    alphas = []
    scores = []

    for i in range(1, alpha + 1):

        reg = Ridge(alpha=i)
        reg.fit(X_train, y_train)

        y_pred = reg.predict(X_test)

        score = r2_score(y_test, y_pred)

        alphas.append(i)
        scores.append(score)

    plt.plot(alphas, scores, marker='o')
    plt.xlabel("Alpha")
    plt.ylabel("R2 Score")
    plt.title("Alpha vs R2 Score")
    plt.show()
```

```python
rdge(500)
```

```python
las = Lasso(alpha=0.005)
las.fit(X_train,y_train)
y_pred = las.predict(X_test)
r2_score(y_test,y_pred)
```

```python
def las(alpha):

    alphas = []
    scores = []

    for i in range(1, alpha + 1):

        las = Lasso(alpha=i)
        las.fit(X_train, y_train)

        y_pred = las.predict(X_test)

        score = r2_score(y_test, y_pred)

        alphas.append(i)
        scores.append(score)

    plt.plot(alphas, scores, marker='o')
    plt.xlabel("Alpha")
    plt.ylabel("R2 Score")
    plt.title("Alpha vs R2 Score")
    plt.show()
```

```python
las(10)
```

```python
Enet = ElasticNet(alpha=0.005,l1_ratio=0.254)
Enet.fit(X_train,y_train)
y_pred = Enet.predict(X_test)
r2_score(y_test,y_pred)
```

```python
def net(alpha, l1_ratio):

    alphas = []
    scores = []

    for i in range(1, alpha + 1):

        model = ElasticNet(alpha=i, l1_ratio=l1_ratio)
        model.fit(X_train, y_train)

        y_pred = model.predict(X_test)

        score = r2_score(y_test, y_pred)

        alphas.append(i)
        scores.append(score)

    plt.plot(alphas, scores, marker='o')
    plt.xlabel("Alpha")
    plt.ylabel("R2 Score")
    plt.title(f"Alpha vs R2 Score (l1_ratio={l1_ratio})")
    plt.show()
```

```python
net(10,0.1)
```

```python
net(10,0.3)
```

```python
net(10,0.9)
```
